## Categorical vs. Numeric Correlation
-----
THe main goal is to see if `kathisma_num` or `bertTopic` is associated with differences in readbilty scores or other metrics. we can just rely on the correlation matrix to interpret these variables. We can use `ANOVA` - Analysis of Variables, to see the realtion between these two cateogrical variables with the rest of the numeric variables. This will show us the existense of any difference the variable have on the readabilty scores

```python
import pingouin as pg
print(pg.__version__)
```

### Metrics
- `Source` - the metric in consideration
- `ddof1` - the numerator degrees of freedom 
     - for 20 Kathismata - $ 20 - 1 = 19$
     - number of groups
- `ddof2` - denominator of degrees of freedom
     - $total_obersvations - #_of_groups$
    
- `F` - the F-statisatic value 
    - <span style="background-color: #ffff00;">The larger the value indicates more of a difference in group means </span>
- `p_unc` - uncorrected p-value for the test
    - <span style="background-color: #ffff00;"> is the ANOVA signifigant</span>
- `np2` - the effect sized measured as partial eta-squared

| Measure     | What you want | Meaning                                                         |
| ----------- | ------------- | --------------------------------------------------------------- |
| **F**       | Larger        | More between-group variation relative to within-group variation |
| **p-value** | **Smaller**   | Stronger evidence that groups differ                            |
| **η²p**     | Larger        | Stronger/larger effect                                          |




```python
scores = ['fk_score', 'gf_score', 'smog_score']
target_metrics = ["kathisma_num", "bertTopic"]

anovas = []

for score in scores:
    for metric in target_metrics:
        result = pg.anova(dv=score, between=metric, data=psalms)

        temp = result[['Source', 'F', 'p_unc', 'np2']].copy()
        temp['score'] = score
        temp['metric'] = metric

        anovas.append(temp)

anovas = pd.concat(anovas, ignore_index=True)

anovas = anovas[['metric', 'score', 'F', 'p_unc', 'np2']]
anovas.sort_values(by='metric')
```

`bertTopic` &rarr; `gf_score` : **F**= 6.04, **p** < .001, **η²p** = .137

This means that GF readability scores differ significantly across the BerTopic groups, with a partial eta-squared of .137.

It does not mean that the topic causes the readability score to change.


```python
posthocs = []

for score in scores:
    for metric in target_metrics:
        # Check ANOVA p-value
        anova_result = pg.anova(
            dv=score,
            between=metric,
            data=psalms
        )

        p_value = anova_result.loc[0, 'p_unc']

        # Only run Tukey if ANOVA is significant
        if p_value < 0.05:
            tukey = pg.pairwise_tukey(
                dv=score,
                between=metric,
                data=psalms
            )

            tukey['score'] = score
            tukey['metric'] = metric

            posthocs.append(tukey)

posthocs = pd.concat(posthocs, ignore_index=True)
```

The p_tukey value is the probability of observing a difference at least this large if the two groups actually had the same population mean.

**Interpretation** 

Using a significance level of α = 0.05:

- **p_tukey** → Is the difference statistically significant?  
    - **p_tukey < 0.05** → the two groups are significantly different.
    - **p_tukey ≥ 0.05** → no significant difference detected.
- **diff** → How large is the mean difference?
- **hedges** → How large is the effect in standardized units? (practical importance)

| Hedges' g | Interpretation  |
| --------- | --------------- |
| 0.00–0.19 | Negligible      |
| 0.20–0.49 | Small           |
| 0.50–0.79 | Medium          |
| 0.80–1.19 | Large           |
| 1.20–1.99 | Very large      |
| ≥ 2.00    | Extremely large |



```python
posthocs = posthocs[
    ['metric', 'score', 'A', 'B', 
     'mean_A', 'mean_B', 'diff', 
     'p_tukey', 'hedges']
]

posthocs = posthocs.rename(columns={'score': 'readabilty_score'})
```

```python
significant_posthocs = posthocs[
    posthocs['p_tukey'] < 0.05
]
significant_posthocs
```

Tukey post-hoc analysis revealed that Kathisma 2 had significantly higher FK scores than Kathisma 16 
- mean difference = 1.66, p = .003, g = 1.82
