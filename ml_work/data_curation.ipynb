## Curating the Data 

```python
psalmSentiment['sentim_psalm_num']
```

```python
psalmsData['psalm_num']
```

```python
cols_to_add = [
    'fk_score',
    'gf_score',
    'smog_score',
    'word_count',
    'avg_word_len',
    'syllable_count',
    'avg_syllables_per_word',
    'polysyllable_count',
    'num_sentences',
    'avg_words_sent',
    'Psalm',
    'kathisma_num',
    'bertTopic'
]

psalmSentiment = psalmSentiment.merge(
    psalmsData[['psalm_num', 'verse'] + cols_to_add],
    left_on=['sentim_psalm_num'],
    right_on=['psalm_num'],
    how='left'
)
```

```python
psalmSentiment.columns
```

```python
# Remove duplicate columns regardless of column name
psalms = psalmSentiment.loc[:, ~psalmSentiment.T.duplicated()]

print(psalms.columns)
```

```python
psalms.drop_duplicates()
psalms.drop(columns=['sentim_verse'], inplace=True)
```

```python
# cleaning up column names
psalms.columns = psalms.columns.str.replace('sentim_', '', regex=True)
psalms.head()
```

```python
psalms.drop(columns=['Psalm', 'vader_score'], inplace=True)

psalms.head()
```

```python
psalms = pd.get_dummies(psalms, columns=['text'], drop_first=True, dtype=int)
psalms = pd.get_dummies(psalms, columns=['vader_category'], drop_first=False, dtype=int)
```

```python
psalms.head()
```

```python
psalms = psalms[['text_Psalter', 'psalm_num', 'verse', 'neg', 'neu', 'pos', 'compound', 'polarity', 'subjectivity', 
        'emotional_score', 'fk_score', 'gf_score', 'smog_score', 'word_count', 
        'avg_word_len', 'syllable_count', 'avg_syllables_per_word', 'polysyllable_count', 
        'num_sentences', 'avg_words_sent', 'kathisma_num', 'bertTopic', 
        'vader_category_Negative', 'vader_category_Neutral', 'vader_category_Positive']]
```

```python
psalms.drop_duplicates(subset=['text_Psalter', 'psalm_num'], inplace=True)
psalms
```

```python
psalms.to_csv("offical_psalms_data.csv")
```
