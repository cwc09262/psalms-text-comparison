## Curating the Data 

```python
psalmSentiment['sentim_psalm_num']
```

```python
psalmsData['psalm_num']
```

```python
cols_to_add = [
    'fk_score',
    'gf_score',
    'smog_score',
    'word_count',
    'avg_word_len',
    'syllable_count',
    'avg_syllables_per_word',
    'polysyllable_count',
    'num_sentences',
    'avg_words_sent',
    'Psalm',
    'kathisma_num',
    'bertTopic'
]

psalmSentiment = psalmSentiment.merge(
    psalmsData[['psalm_num', 'verse'] + cols_to_add],
    left_on=['sentim_psalm_num'],
    right_on=['psalm_num'],
    how='left'
)
```
